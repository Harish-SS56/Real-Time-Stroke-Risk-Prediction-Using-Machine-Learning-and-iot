```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
from matplotlib.cm import rainbow

import warnings
warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

```python
dataset = pd.read_csv('healthcare-dataset-stroke-data(3).csv')
```

```python
dataset.info()
```

```python
print('Finding missing values')
print(dataset.isnull().sum())
```

```python
#dataset['id']=dataset['id'].dataset(dataset['id'].mean())
dataset['bmi']=dataset['bmi'].fillna(dataset['bmi'].mean()) 
#dataset
```

```python
print('Finding missing values')
print(dataset.isnull().sum())
```

```python
dataset = dataset.dropna()
dataset
```

```python
from sklearn.preprocessing import LabelEncoder

labelencoder_X = LabelEncoder()

dataset['gender'] = labelencoder_X.fit_transform(dataset['gender'])
dataset['ever_married'] = labelencoder_X.fit_transform(dataset['ever_married'])
dataset['work_type'] = labelencoder_X.fit_transform(dataset['work_type'])
dataset['Residence_type'] = labelencoder_X.fit_transform(dataset['Residence_type'])
dataset['smoking_status'] = labelencoder_X.fit_transform(dataset['smoking_status'])

```

```python
dataset
```

```python
y = dataset['stroke']
X = dataset.drop(['stroke'], axis = 1)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.35, random_state = 0)
```

```python
gnb = GaussianNB()
```

```python
#Train the model using the training sets
gnb.fit(X_train, y_train)
```

```python
X_test
```

```python
#Predict the response for test dataset
y_pred = gnb.predict(X_test)
for i in range(0, len(y_pred)):
    i2=str(i)
    res=''
    if y_pred[i]==0:
        res='normal'
    elif y_pred[i]==1:
        res='Stroke'
    print('Test data ',i2 , ' is ',res)
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
from sklearn import metrics
print("Accuracy for Naive Bayes Classifier : {}%".format(metrics.accuracy_score(y_test, y_pred)*100))
```
